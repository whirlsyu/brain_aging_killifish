$$
v = \sqrt{(v_x)^2 + (v_y)^2}
$$

```python
# velocity average per sample
import glob
import os
import csv
import math

csv_files_02 = glob.glob('trajectories_tidy/*_02*.csv')

for file in csv_files_02:
    va = []
    with open(file, "r") as f:
        reader = csv.DictReader(f)
        for line in reader:
            if int(line["frame"]) > 1500*30 or int(line["frame"]) < 1200*30:
                continue
            vx = float(line["velocity_x"])
            vy = float(line["velocity_y"])
            v = math.sqrt(vx**2 + vy**2)
            va.append(v)
    ve = sum(va) / len(va)
    print(f"{os.path.basename(file).replace("fish_video_", "").replace(".csv", "").replace("_1", "")}\t{ve}")

csv_files_07 = glob.glob('trajectories_tidy/*_07*.csv')

for file in csv_files_07:
    va = []
    with open(file, "r") as f:
        reader = csv.DictReader(f)
        for line in reader:
            if int(line["frame"]) > 1500*30 or int(line["frame"]) < 1200*30:
                continue
            vx = float(line["velocity_x"])
            vy = float(line["velocity_y"])
            v = math.sqrt(vx**2 + vy**2)
            va.append(v)
    ve = sum(va) / len(va)
    print(f"{os.path.basename(file).replace("fish_video_", "").replace(".csv", "").replace("_1", "")}\t{ve}")

```
